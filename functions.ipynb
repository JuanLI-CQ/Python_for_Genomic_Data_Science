```python
# A function to compute the GC percentage of a DNA sequence
# Scope of variable declaration
# If you declare a varable inside a function it will only exist in that function

# input: DNA sequence
# output: GC percentage
def gc(dna):
    "this function computes the GC percentage of a dna sequence"
    nbases=dna.count('n')+dna.count('N')
    gcpercent=float(dna.count('c')+dna.count('C')+dna.count('g')+dna.count('G'))*100.0/(len(dna)-nbases)
    return gcpercent

gc('AAGTNNAGTCC')
```

```python
# Boolean functions
# Boolean functions are functions that reture a True or False value. They can be used in conditonals such as if or while statements whenever
# a conditon it too complex

# Write a program that chceks if a given DNA sequence contains an in-frame stop codon.

#!/usr/bin/python3

# define has_stop_codon function here
dna=input("Enter a DNA sequence, please:")

def has_stop_codon(dna):
    "This function checks if given DNA sequence has in-frame stop codons."
    stop_codon_found=False
    stop_codons=['tga','tag','taa']
    for i in range(0,len(dna),3):
        codon=dna[i:i+3].lower()
        if codon in stop_codons:
            stop_codon_found=True
            break
    return stop_codon_found

if(has_stop_codon(dna)):
    print("Input sequence has an in-frame stop codon.")
else:
    print("Input sequence has no in-frame stop codon.")
```

```python
# Define function defalt values
# Suppose the has_stop_codon function alse accepts a frame argument (equal to 0,1,or 2) which specifies in what fream 
# we want to look for stop codons.
def has_stop_codon(dna,frame):
    "This function checks if given DNA sequence has in-frame stop codons."
    stop_codon_found=False
    stop_codons=['tga','tag','taa']
    for i in range(frame,len(dna),3):
        codon=dna[i:i+3].lower()
        if codon in stop_codons:
            stop_codon_found=True
            break
    return stop_codon_found

dna='atgagcggccggct'
has_stop_codon(dna,0)
has_stop_codon(dna,1)
has_stop_codon(dna,2)
```

```python
# We would like the has_stop_codon() function to check for in-frame stop codons by default, 
# i.e. if we call the function like this:
# has_stop_codon(dna), to assume that frame is 0

# Use the 0 value as the default value for the frame parameter
def has_stop_codon(dna,frame=0):
    "This function checks if given DNA sequence has in-frame stop codons."
    stop_codon_found=False
    stop_codons=['tga','tag','taa']
    for i in range(frame,len(dna),3):
        codon=dna[i:i+3].lower()
        if codon in stop_codons:
            stop_codon_found=True
            break
    return stop_codon_found

dna='aaatgagcggccggct'
has_stop_codon(dna)
has_stop_codon(dna,1)
```

```python
# Passing function arguments
# Passing arguments by position
seq='tgggcctaggtaac'
has_stop_codon(seq,1)
```

```python
# It is also possible to pass arguments by name
has_stop_codon(frame=0, dna=seq)
```

```python
# A function to reverse a string
# Regular slices
dna='AGTGTGGGGCG'
print(dna[0:3])
# Extended slices, get string from postion 1 to 3, and skip every string at 2 postion (third number is step argument). 
print(dna[0:3:2])
# provide colons with no arguments, it gives us the whole slices of string, 
print(dna[::])
# then add step argument -1, which mean every one step at a time but backwards
print(dna[::-1])

# A function to reverse a string
def reverse_string(seq):
    seq=seq[::-1]
    return seq 
```

```python
# A function to complement a DNA sequence
def complement(dna):
    """Return the complementary sequence string."""
    # creat a complement dictionary that give complementary bases
    basecomplement = {'A':'T','C':'G','G':'C','T':'A','N':'N',
                      'a':'t','c':'g','g':'c','t':'a','n':'n'}
    # create a list by transforming dna sequence to a list and each element in the string is transformed to a element in the new list "letters"
    letters = list(dna)
    # get complementary base of each base element in the list letters using the base dictionary, 
    # use base as key to get value base it's corresponding to
    letters = [basecomplement[base] for base in letters]

    # change list back to string using join function
    return ''.join(letters)

# List comprehensions
# List comprehensions in Python provide a concise way to create lists
# Common applications are to make new lists where each element is the result of some operations applied to each memeber of another sequence, 
# or to create a subsequence of those elements that satisfy a certain condition.

# Syntax
# new_list = [operation(i) for i in old_list if filter(i)] ## it is equivalent to below commands

# new_list = []
# for i in old_list:
#     if filter(i):
#         new_list.append(operation(i))

# Split and join
# Split and join are methods of the string object.
# Split: the method split() returns a list of all the words in the string:
sentence='enzymes and other proteins come in many shapes'
print(sentence.split())

# By default it splits on all white spaces, but we can specify a separator
print(sentence.split('and'))

# Join: The method join() returns a string in which the string elements were joined from a list. 
# The separator string that joins the elements is the one upon which the function is called
print('-'.join(['enzymes', 'and', 'other', 'proteins', 'come', 'in', 'many', 'shapes']))
    
```

```python
# A more complex function example
# Write a function to reverse complement a DNA sequence.
def reverse_complement(seq):
    """Return the reverse complement of the dna string."""
    # pseudocode here
    seq=reverse_string(seq)
    seq=complement(seq)
    
    return seq

reverse_complement('CCGGAAGACGTTACTTAG')
```
